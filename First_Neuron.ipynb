```python
import torch
import torch.nn as nn
```

```python
torch.manual_seed(42)
inputs=torch.tensor([1.2,1.3,2.8])
```

```python
neuron=nn.Linear(in_features=3,out_features=1)
```

```python
output=neuron(inputs)
```

```python
output
```

```python
#Verify using formula.....
neuron.weight @ inputs + neuron.bias

#WE are geeting same output but if we run agaain we may get different output as bias value is taken randomly and that can change the output
# to make the output same while keeping the same bias we will use maunal seed
```
